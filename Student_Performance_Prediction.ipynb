```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.linear_model import LinearRegression
lr2 = LinearRegression()
```

```python
data=pd.read_csv('student_habits_performance.csv')
```

```python
data.head()
```

```python
data.drop(['parental_education_level','diet_quality','internet_quality'],axis=1,inplace=True)
```

```python
data.gender = data.gender.apply(lambda x: 1 if x=='Female' else 0)
data.part_time_job = data.part_time_job.apply(lambda x: 1 if x=='Yes' else 0)
data.extracurricular_participation = data.extracurricular_participation.apply(lambda x: 1 if x=='Yes' else 0)
```

```python
data['student_id']=data['student_id'].str.replace('S','').astype(int)
```

```python
data.corr()
```

```python
X=data.drop('exam_score',axis=1)
y=data['exam_score']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
lr2.fit(X_train,y_train)
```

```python
y_pred_train = lr2.predict(X_train)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mean_squared_error(y_train, y_pred_train)
```

```python
r2_score(y_train,y_pred_train)
```

```python
lr2.score(X_test,y_test)
```
